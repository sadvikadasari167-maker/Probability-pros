```python
# ============================================================
# 2D GAME TRANSFORMATION ENGINE
# B.Tech Mathematics Mini Project - F3
# ============================================================

"""
Mathematical model:

A 2D point is represented using homogeneous coordinates:

        [ x ]
P   =   [ y ]
        [ 1 ]

Transformations:

Scaling:
        [ s  0  0 ]
S   =   [ 0  s  0 ]
        [ 0  0  1 ]

Rotation:
        [ cos(theta)  -sin(theta)  0 ]
R   =   [ sin(theta)   cos(theta)  0 ]
        [     0            0        1 ]

Translation:
        [ 1  0  tx ]
T   =   [ 0  1  ty ]
        [ 0  0   1 ]

Combined transformation:

        M = T @ R @ S

For every vertex:

        P_final = M @ P_original
"""
```

```python
# ============================================================
# CELL 2: IMPORTS
# ============================================================

import numpy as np
import matplotlib.pyplot as plt

# Optional: SymPy for symbolic mathematics
import sympy as sp

# Make NumPy output easier to read
np.set_printoptions(precision=4, suppress=True)

print("Libraries imported successfully.")
```

```python
# ============================================================
# CELL 3: INPUT CONFIGURATION
# ============================================================

# Triangle vertices: (x, y)
# Each row represents one vertex.
triangle = np.array([
    [1.0, 1.0],   # Vertex A
    [4.0, 1.0],   # Vertex B
    [2.0, 3.0]    # Vertex C
])

# Transformation inputs
rotation_angle_deg = 30.0   # Rotation angle in degrees
scale_factor = 2.0          # Uniform scaling factor

translation_x = 5.0         # Translation in x direction
translation_y = 3.0         # Translation in y direction

print("Original Triangle:")
print(triangle)

print("\nTransformation Inputs:")
print("Rotation angle :", rotation_angle_deg, "degrees")
print("Scale factor   :", scale_factor)
print("Translation X  :", translation_x)
print("Translation Y  :", translation_y)
```

```python
# ============================================================
# CELL 4: CONSTRUCT TRANSFORMATION MATRICES
# ============================================================

# Convert degrees to radians because NumPy's sin/cos
# functions use radians.
theta = np.deg2rad(rotation_angle_deg)

# ------------------------------------------------------------
# 1. Scaling matrix
# ------------------------------------------------------------

S = np.array([
    [scale_factor, 0, 0],
    [0, scale_factor, 0],
    [0, 0, 1]
], dtype=float)


# ------------------------------------------------------------
# 2. Rotation matrix
# ------------------------------------------------------------

R = np.array([
    [np.cos(theta), -np.sin(theta), 0],
    [np.sin(theta),  np.cos(theta), 0],
    [0, 0, 1]
], dtype=float)


# ------------------------------------------------------------
# 3. Translation matrix
# ------------------------------------------------------------

T = np.array([
    [1, 0, translation_x],
    [0, 1, translation_y],
    [0, 0, 1]
], dtype=float)


print("SCALING MATRIX (S)")
print(S)

print("\nROTATION MATRIX (R)")
print(R)

print("\nTRANSLATION MATRIX (T)")
print(T)
```

```python
# ============================================================
# CELL 5: STEP-BY-STEP TRANSFORMATION
# ============================================================

# Convert triangle coordinates into homogeneous coordinates.
#
# Original:
# [x, y]
#
# Homogeneous:
# [x]
# [y]
# [1]

ones = np.ones((triangle.shape[0], 1))

triangle_h = np.hstack([triangle, ones]).T


# ------------------------------------------------------------
# STEP 1: Scaling
# ------------------------------------------------------------

triangle_scaled = S @ triangle_h


# ------------------------------------------------------------
# STEP 2: Rotation
# ------------------------------------------------------------

triangle_rotated = R @ triangle_scaled


# ------------------------------------------------------------
# STEP 3: Translation
# ------------------------------------------------------------

triangle_final = T @ triangle_rotated


# Convert results back from homogeneous coordinates
# to ordinary (x, y) coordinates.

scaled_xy = triangle_scaled[:2, :].T
rotated_xy = triangle_rotated[:2, :].T
final_xy = triangle_final[:2, :].T


print("ORIGINAL TRIANGLE")
print(triangle)

print("\nAFTER SCALING")
print(scaled_xy)

print("\nAFTER ROTATION")
print(rotated_xy)

print("\nAFTER TRANSLATION / FINAL")
print(final_xy)
```

```python
# ============================================================
# CELL 6: TRANSFORMATION COMPOSITION
# ============================================================

# Required composition:
#
# M = T × R × S
#
# Using NumPy's @ operator for matrix multiplication.

M = T @ R @ S

print("TOTAL TRANSFORMATION MATRIX")
print("M = T × R × S")
print(M)


# Verify that applying M directly gives the same result
# as applying S, then R, then T.

triangle_direct = M @ triangle_h

direct_xy = triangle_direct[:2, :].T

print("\nFINAL COORDINATES USING TOTAL MATRIX")
print(direct_xy)


# Numerical verification
if np.allclose(final_xy, direct_xy):
    print("\n✓ Verification successful!")
    print("Step-by-step transformation = Combined matrix transformation")
else:
    print("\n✗ Verification failed.")
```

```python
# ============================================================
# CELL 7: FORMATTED OUTPUT TABLE
# ============================================================

print("=" * 70)
print("2D GAME TRANSFORMATION ENGINE - RESULTS")
print("=" * 70)

print(f"\nRotation angle : {rotation_angle_deg}°")
print(f"Scale factor   : {scale_factor}")
print(f"Translation    : ({translation_x}, {translation_y})")

print("\n" + "-" * 70)
print(f"{'Vertex':<10}{'Original':<20}{'Scaled':<20}{'Rotated':<20}{'Final'}")
print("-" * 70)

labels = ["A", "B", "C"]

for i, label in enumerate(labels):
    print(
        f"{label:<10}"
        f"({triangle[i,0]:.2f}, {triangle[i,1]:.2f})"
        f"{'':<8}"
        f"({scaled_xy[i,0]:.2f}, {scaled_xy[i,1]:.2f})"
        f"{'':<8}"
        f"({rotated_xy[i,0]:.2f}, {rotated_xy[i,1]:.2f})"
        f"{'':<8}"
        f"({final_xy[i,0]:.2f}, {final_xy[i,1]:.2f})"
    )

print("-" * 70)
```

```python
# ============================================================
# CELL 8: STEP-BY-STEP VISUALIZATION
# ============================================================

def close_polygon(points):
    """Close the triangle so that the final edge is drawn."""
    return np.vstack([points, points[0]])


fig, ax = plt.subplots(figsize=(9, 7))

# Original
p = close_polygon(triangle)
ax.plot(p[:, 0], p[:, 1], marker='o', label='Original')

# Scaled
p = close_polygon(scaled_xy)
ax.plot(p[:, 0], p[:, 1], marker='o', label='After Scaling')

# Rotated
p = close_polygon(rotated_xy)
ax.plot(p[:, 0], p[:, 1], marker='o', label='After Rotation')

# Final
p = close_polygon(final_xy)
ax.plot(p[:, 0], p[:, 1], marker='o', label='Final')

# Label original vertices
for i, label in enumerate(["A", "B", "C"]):
    ax.annotate(
        label,
        (triangle[i, 0], triangle[i, 1]),
        xytext=(5, 5),
        textcoords="offset points"
    )

ax.axhline(0, linewidth=0.8)
ax.axvline(0, linewidth=0.8)

ax.set_title("2D Game Transformation Engine")
ax.set_xlabel("X coordinate")
ax.set_ylabel("Y coordinate")

ax.grid(True, alpha=0.3)
ax.legend()

plt.show()
```

```python
# ============================================================
# CELL 9: VISUALIZE THE TOTAL MATRIX
# ============================================================

fig, ax = plt.subplots(figsize=(7, 4))

ax.imshow(np.abs(M))

# Display matrix values
for i in range(3):
    for j in range(3):
        ax.text(
            j,
            i,
            f"{M[i, j]:.3f}",
            ha="center",
            va="center"
        )

ax.set_xticks(range(3))
ax.set_yticks(range(3))

ax.set_xticklabels(["x", "y", "translation"])
ax.set_yticklabels(["x'", "y'", "homogeneous"])

ax.set_title("Total Transformation Matrix M = T × R × S")

plt.show()
```

```python
# ============================================================
# CELL 10: SYMBOLIC MATHEMATICAL VERIFICATION
# ============================================================

s, tx, ty, theta = sp.symbols('s tx ty theta')

S_sym = sp.Matrix([
    [s, 0, 0],
    [0, s, 0],
    [0, 0, 1]
])

R_sym = sp.Matrix([
    [sp.cos(theta), -sp.sin(theta), 0],
    [sp.sin(theta),  sp.cos(theta), 0],
    [0, 0, 1]
])

T_sym = sp.Matrix([
    [1, 0, tx],
    [0, 1, ty],
    [0, 0, 1]
])

M_sym = sp.simplify(T_sym * R_sym * S_sym)

print("Symbolic Total Transformation Matrix:")
sp.pprint(M_sym)
```
